```python
import pandas as pd
```

```python
df = pd.read_csv(r"customer_shopping_behavior.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
median_values = df.groupby("Category")["Review Rating"].transform("median")
df["Review Rating"] = df["Review Rating"].fillna(median_values)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns= {"purchase_amount_(usd)" : "purchase_amount"})
```

```python
### Create a new column age_group

df.loc[(df["age"] >= 18) & (df["age"] < 25), "age_group"] = "Young"
df.loc[(df["age"] >= 25) & (df["age"] < 40), "age_group"] = "Adult"
df.loc[(df["age"] >= 40) & (df["age"] < 59), "age_group"] = "Middle Age"
df.loc[(df["age"] >= 60) & (df["age"]), "age_group"] = "Senior"
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days
df['frequency_of_purchases'].unique()
```

```python
# df.loc[df['frequency_of_purchases'] == 'Fortnightly','purchases_frequency_days'] = 14
mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30,
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']]
```

```python
df.info()
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# removing 'promo_code_used' because 'discount_applied' contains the same value.

df = df.drop('promo_code_used',axis=1)
```
